## 🚀 Cell 4 — Launch the Web App

This will:
1. Start Streamlit in the background
2. Open a public tunnel via localtunnel
3. Print a `https://xxx.loca.lt` URL

**Click that URL** → the loca.lt warning page appears → click **"Click to Continue"** → paste the IP from Cell 3 as the password → the app opens.

**The first model load takes ~30 seconds** (loading EfficientNet + YOLO + Faster R-CNN into GPU).

```python
# Kill any previous Streamlit process (safe to re-run)
!pkill -f streamlit 2>/dev/null
import time
time.sleep(2)

# Launch Streamlit in the background, log to file
!streamlit run /content/drive/MyDrive/CV_project_v4/fracture_app.py \
    --server.port 8501 \
    --server.headless true \
    --server.fileWatcherType none \
    --browser.gatherUsageStats false \
    &>/content/streamlit_log.txt &

# Wait for Streamlit to actually start
print('⏳ Waiting for Streamlit to start...')
time.sleep(10)
print('✅ Streamlit started\n')
print('=' * 60)
print('  Opening public tunnel — your URL appears below 👇')
print('  Click it, paste the IP from Cell 3 as the password')
print('=' * 60)

# Open localtunnel — this will print the public URL and stay running
!npx --yes localtunnel --port 8501
```

## 💡 Troubleshooting

| Problem | Fix |
|---|---|
| `ModuleNotFoundError` | Re-run Cell 2 |
| App says "Missing model files" | Check that all 4 checkpoints are in `CV_project_v4/` on Drive |
| Tunnel URL gives 502 | Wait 10 more seconds, then refresh. Streamlit may still be loading. |
| Localtunnel password rejected | Re-run Cell 3, use the **new** IP it prints |
| GPU not detected | Runtime → Change runtime type → T4 GPU → Save → re-run from Cell 1 |
| Drive not mounted error | Re-run Cell 1 |
| Want to restart cleanly | Cell 6 → Cell 4 |

**Note:** Localtunnel passwords change with each session. If the IP changes (rare), re-run Cell 3.
